```python
from ipywidgets import interact, interactive, fixed, interact_manual
import ipywidgets as widgets
from IPython.display import display, Image

import os
```

```python
# Get image files
image_folder = "../images/saf_skyrmion-results-0.8nm"
image_files = [f for f in os.listdir(image_folder) if f.lower().endswith(('.png', '.jpg', '.jpeg', '.gif', '.bmp'))]

# Create dropdown
dropdown = widgets.Dropdown(
    options=image_files,
    value=image_files[0],
    description='Select Image:',
)

# Interactive function to display selected image
def show_image(image_name):
    return Image(os.path.join(image_folder, image_name))

# Use interact to link dropdown with display function
widgets.interact(show_image, image_name=dropdown)
```

```python
# Get image files
image_folder = "../images/saf_skyrmion-results-0.8nm"
image_files = {
    f: f"D={f.split('-')[1]} and Ms={f.split('-')[2].replace('_','/')}" 
    for f in os.listdir(image_folder) 
    if f.lower().endswith(('.png', '.jpg', '.jpeg', '.gif', '.bmp'))
}

# Create dropdown
dropdown = widgets.Dropdown(
    options=image_files,
    value='skyrmion-150nm-260kA_m.png',
    description='Select Image:'
)

# Interactive function to display selected image
def show_image(image_name):
    return Image(os.path.join(image_folder, image_name))

# Use interact to link dropdown with display function
widgets.interact(show_image, image_name=dropdown)
```
